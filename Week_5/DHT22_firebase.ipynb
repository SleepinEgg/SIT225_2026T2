```python
from firebase_admin import db
import serial
import time
from datetime import datetime, timezone

# A reference point is always needed to be set
# before any operation is carried out on a database.

ref_dht22 = db.reference("/sensors/dht22")

ser = serial.Serial('COM3', 9600) 
time.sleep(2)


# Read 3 real-time samples from sensor
readings_count = 0
while readings_count < 3:
    line = ser.readline().decode().strip().split(',')
    
    if len(line) == 2:
        data = {
            "timestamp": datetime.now().isoformat(),
            "humidity": float(line[0]),
            "temperature": float(line[1])
        }
        ref_dht22.push(data)
        readings_count += 1

ser.close()

all_dht_data = ref_dht22.get()

for key, value in all_dht_data.items():
    print(f"ID: {key} | Temp: {value['temperature']}°C | Humidity: {value['humidity']}% | Time: {value['timestamp']}")
```

```python
from firebase_admin import db
import serial
import time
from datetime import datetime, timezone

# A reference point is always needed to be set
# before any operation is carried out on a database.

ref_sr04 = db.reference("/sensors/sr04")

ser = serial.Serial('COM3', 9600) 
time.sleep(2)


# Read 3 real-time samples from sensor
readings_count = 0
while readings_count < 3:
    line = ser.readline().decode().strip()
    
    if line:
        data = {
            "timestamp": datetime.now().isoformat(),
            "distance_cm": float(line),
        }
        ref_sr04.push(data)
        readings_count += 1

ser.close()

all_sr04_data = ref_sr04.get()

for key, value in all_sr04_data.items():
    print(f"ID: {key} | Distance: {value['distance_cm']} cm | Time: {value['timestamp']}")
```
