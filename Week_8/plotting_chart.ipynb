```python
data_x['time'] = pd.to_datetime(data_x['time'])
data_y['time'] = pd.to_datetime(data_y['time'])
data_z['time'] = pd.to_datetime(data_z['time'])
data_linear['time'] = pd.to_datetime(data_linear['time'])


data_x['seconds'] = (data_x['time'] - data_x['time'].iloc[0]).dt.total_seconds()
data_y['seconds'] = (data_y['time'] - data_y['time'].iloc[0]).dt.total_seconds()
data_z['seconds'] = (data_z['time'] - data_z['time'].iloc[0]).dt.total_seconds()
data_linear['seconds'] = (data_linear['time'] - data_linear['time'].iloc[0]).dt.total_seconds()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(data_x['time'], data_x['value'], label='X-axis', color='blue')
plt.title('Accelerometer Value X')
plt.figure(figsize=(12, 6))
plt.plot(data_y['time'], data_y['value'], label='Y-axis', color='orange')
plt.title('Accelerometer Value Y')
plt.figure(figsize=(12, 6))
plt.plot(data_z['time'], data_z['value'], label='Z-axis', color='green')
plt.title('Accelerometer Value Z')
plt.figure(figsize=(12, 6))
plt.plot(data_linear['time'], data_linear['value'], label='Linear Acceleration', color='red')
plt.title('Accelerometer Data Linear')
plt.xlabel('Time')
```
